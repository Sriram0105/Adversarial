```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# --------------------------
# Load and preprocess DNS dataset
# --------------------------
def load_dns_data():
    legit = pd.read_csv("adv-ml-defence/dataset/legitimate.csv")
    cw = pd.read_csv("adv-ml-defence/dataset/cw_combined.csv")
    fgsm = pd.read_csv("adv-ml-defence/dataset/fgsm_combined.csv")
    jsma = pd.read_csv("adv-ml-defence/dataset/jsma_combined.csv")

    legit["label"] = 0
    cw["label"] = 1
    fgsm["label"] = 1
    jsma["label"] = 1

    combined = pd.concat([legit, cw, fgsm, jsma], ignore_index=True)
    combined = combined.sample(frac=1, random_state=42).reset_index(drop=True)

    X = combined.drop(columns=["label"])
    y = combined["label"]

    scaler = StandardScaler()
    X_scaled = scaler.fit_transform(X)

    return train_test_split(X_scaled, y, test_size=0.3, random_state=42), scaler

# --------------------------
# Victim model: now softmax 2-class output
# --------------------------
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Input
from tensorflow.keras.optimizers import Adam

def build_model(input_dim):
    model = Sequential([
        Input(shape=(input_dim,)),
        Dense(512, activation='relu'),
        Dense(256, activation='relu'),
        Dense(128, activation='relu'),
        Dense(2, activation='softmax')  # softmax for 2 classes
    ])
    model.compile(optimizer=Adam(1e-4), loss='sparse_categorical_crossentropy', metrics=['accuracy'])
    return model

# --------------------------
# Train victim model + generate attacks
# --------------------------
from art.estimators.classification import TensorFlowV2Classifier
from art.attacks.evasion import FastGradientMethod, CarliniL2Method, SaliencyMapMethod
import tensorflow as tf

def generate_attacks(X_train, y_train, X_test):
    input_dim = X_train.shape[1]
    model = build_model(input_dim)

    model.fit(X_train, y_train, epochs=10, batch_size=32, verbose=1)

    classifier = TensorFlowV2Classifier(
        model=model,
        nb_classes=2,
        input_shape=(input_dim,),
        loss_object=tf.keras.losses.SparseCategoricalCrossentropy(),
        clip_values=(X_train.min(), X_train.max())
    )

    # FGSM
    fgsm = FastGradientMethod(estimator=classifier, eps=0.1)
    X_fgsm = np.nan_to_num(fgsm.generate(x=X_test))

    # CW (now works!)
    cw = CarliniL2Method(classifier=classifier, max_iter=10)
    X_cw = np.nan_to_num(cw.generate(x=X_test))

    # JSMA (targeted attack — we'll just flip class)
    y_target = 1 - y_test[:len(X_test)]  # flip 0->1 or 1->0
    jsma = SaliencyMapMethod(classifier=classifier)
    X_jsma = np.nan_to_num(jsma.generate(x=X_test, y=y_target))

    return X_fgsm, X_cw, X_jsma, classifier.model

# --------------------------
# Load data and generate attacks
# --------------------------
(X_train, X_test, y_train, y_test), scaler = load_dns_data()
X_fgsm, X_cw, X_jsma, victim_model = generate_attacks(X_train, y_train, X_test)

# --------------------------
# Combine and clean
# --------------------------
X_combined = np.vstack((X_test, X_fgsm, X_cw, X_jsma))
y_combined = np.concatenate((
    np.zeros(len(X_test)),      # Clean
    np.ones(len(X_fgsm)),       # FGSM
    np.ones(len(X_cw)),         # CW
    np.ones(len(X_jsma))        # JSMA
))
X_combined = np.nan_to_num(X_combined)

# --------------------------
# Train a logistic regression detector
# --------------------------
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import confusion_matrix, classification_report

detector = LogisticRegression()
detector.fit(X_combined, y_combined)
y_pred = detector.predict(X_combined)

print("\n📊 Confusion Matrix:")
print(confusion_matrix(y_combined, y_pred))

print("\n📈 Classification Report:")
print(classification_report(y_combined, y_pred))
```

```python
from sklearn.utils import resample

# Separate classes
X_clean = X_combined[y_combined == 0]
X_adv = X_combined[y_combined == 1]

# Downsample adversarial to match clean
X_adv_down, y_adv_down = resample(X_adv, np.ones(len(X_adv)), n_samples=len(X_clean), random_state=42)

# Combine balanced dataset
X_balanced = np.vstack((X_clean, X_adv_down))
y_balanced = np.concatenate((np.zeros(len(X_clean)), y_adv_down))

# Shuffle
from sklearn.utils import shuffle
X_balanced, y_balanced = shuffle(X_balanced, y_balanced, random_state=42)
```

```python
import torch
import torch.nn as nn
import numpy as np

def fgsm_attack(model, loss_fn, X, y, epsilon=0.1):
    """
    Generate adversarial examples using FGSM.
    - model: PyTorch model
    - loss_fn: Loss function (e.g., nn.CrossEntropyLoss())
    - X: Input features (numpy array)
    - y: True labels (numpy array)
    - epsilon: Perturbation strength
    """
    # Convert to tensors
    X_tensor = torch.tensor(X, dtype=torch.float32, requires_grad=True)
    y_tensor = torch.tensor(y, dtype=torch.long)

    # Forward pass
    outputs = model(X_tensor)
    loss = loss_fn(outputs, y_tensor)

    # Backward pass
    model.zero_grad()
    loss.backward()

    # FGSM perturbation: sign of gradient
    sign_grad = X_tensor.grad.data.sign()
    X_adv = X_tensor + epsilon * sign_grad

    # Return perturbed samples as NumPy
    return X_adv.detach().numpy()
```

```python
# Prepare model (simple example)
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

from torch import nn
from torch.utils.data import DataLoader, TensorDataset

# Example classifier
class Net(nn.Module):
    def __init__(self, in_dim):
        super(Net, self).__init__()
        self.fc = nn.Sequential(
            nn.Linear(in_dim, 32),
            nn.ReLU(),
            nn.Linear(32, 2)
        )
    
    def forward(self, x):
        return self.fc(x)

# Train simple model (you may skip if model already exists)
X_scaled = scaler.transform(X_test)  # ensure scaling
y = y_test[:len(X_scaled)]

model = Net(X_scaled.shape[1])
criterion = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(model.parameters(), lr=0.01)

X_tensor = torch.tensor(X_scaled, dtype=torch.float32)
y_tensor = torch.tensor(y, dtype=torch.long)

dataset = TensorDataset(X_tensor, y_tensor)
loader = DataLoader(dataset, batch_size=32, shuffle=True)

for epoch in range(10):
    for xb, yb in loader:
        optimizer.zero_grad()
        output = model(xb)
        loss = criterion(output, yb)
        loss.backward()
        optimizer.step()

# Generate FGSM adversarial samples
X_fgsm = fgsm_attack(model, criterion, X_scaled, y)
```

```python
from scipy.stats import ks_2samp
import numpy as np

def detect_changed_features(X_clean, X_adv, feature_names=None):
    changed = []

    for i in range(X_clean.shape[1]):
        stat, p_value = ks_2samp(X_clean[:, i], X_adv[:, i])
        changed.append((i, stat, p_value))

    # Sort by KS statistic (higher = more different)
    changed.sort(key=lambda x: x[1], reverse=True)

    print(f"{'Feature':<8} {'KS Statistic':<15} {'P-value'}")
    print("=" * 35)
    for i, stat, p in changed:
        name = feature_names[i] if feature_names else f"F{i}"
        print(f"{name:<8} {stat:<15.4f} {p:.4e}")

    return changed
feature_names = [f"F{i}" for i in range(X_test.shape[1])]
detect_changed_features(X_test, X_fgsm, feature_names)
```
